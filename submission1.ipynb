Find difference between pyhton list and numpy array.

| Feature                     | Python List                                                   | NumPy Array                                  |
| --------------------------- | ------------------------------------------------------------- | -------------------------------------------- |
| **Definition**              | Built-in data structure in Python                             | Data structure provided by **NumPy** library |
| **Data type**               | Can store **different data types** (int, float, string, etc.) | Stores **same data type only**               |
| **Speed**                   | Slower for numerical operations                               | Much **faster** for numerical calculations   |
| **Memory usage**            | Uses **more memory**                                          | Uses **less memory**                         |
| **Mathematical operations** | Does **not support** direct element-wise operations           | Supports **element-wise operations**         |
| **Multidimensional data**   | Limited support                                               | Excellent support (1D, 2D, 3D, etc.)         |
| **Usage**                   | General-purpose programming                                   | Scientific computing, data analysis, ML      |
| **Library required**        | No extra library needed                                       | Requires **NumPy**                           |


Python lists can store multiple data types and are slower for numerical operations, while NumPy arrays store homogeneous data and provide faster, memory-efficient numerical computation.

1. Why numpy is faster then list?

    NumPy arrays store elements of the same data type (homogeneous), so no type checking is needed.

    Data is stored in continuous memory locations, improving access speed.  

    NumPy uses optimized C and Fortran code internally for calculations.

    Supports vectorized operations, avoiding slow Python loops.
